```python
import PyPDF2
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
pdf_file = list(uploaded.keys())[0]
reader = PyPDF2.PdfReader(pdf_file)
text = ""
for page in reader.pages:
  page_text = page.extract_text()
  if page_text:
    text += page_text + " "
    print(text[:1000])
```

```python
def create_chunks(text,chunk_size=500):
  words = text.split()
  chunks = []

  for i in range(0,len(words), chunk_size):
    chunk = " ".join(words[i:i+chunk_size])
    chunks.append(chunk)

  return chunks

chunks = create_chunks(text)
print("Total chunks:",len(chunks))
```

```python
def answer_question(question, chunks):
    vectorizer = TfidfVectorizer(stop_words='english')
    documents = chunks + [question]

    tfidf_matrix = vectorizer.fit_transform(documents)

    question_vector = tfidf_matrix[-1]
    chunk_vectors = tfidf_matrix[:-1]

    similarities = cosine_similarity(question_vector, chunk_vectors)
    best_match_index = similarities.argmax()
    similarity_score = similarities[0][best_match_index]

    return chunks[best_match_index], similarity_score
```

```python
question = input("Ask a question from your notes: ")

answer, score = answer_question(question, chunks)

print("Question:", question)
print("\nMost Relevant Answer:")
print(answer)
print("\nSimilarity Score:", round(score, 2))
```

# **## Project Workflow**

1. Upload PDF notes
2. Extract text from PDF
3. Split text into chunks
4. Convert text into TF-IDF vectors
5. Compare user question with chunks using cosine similarity
6. Return the most relevant answer
